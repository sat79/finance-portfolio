# Lesson 01 — Your first finance calculation in Python

### What you will learn
1. What a **variable** is (a name that stores a value).
2. What a **function** is (reusable code that performs a task).
3. How Python calculates revenue growth and operating profit margin.
4. Why a trustworthy AI-finance system needs checked calculations, not guessed numbers.

> Important: This is **not an AI agent yet**. We are building the reliable numerical foundations that a later AI agent will use.

All values refer to a **fictional company** and are measured in **£ million**.

```python
# Step 1: Store the fictional company's numbers in variables.
company = "Example Retail Group (fictional)"
revenue_2023 = 1100.0
revenue_2024 = 1250.0
operating_profit_2023 = 140.0
operating_profit_2024 = 175.0

print(company)
print("2024 revenue (£m):", revenue_2024)
```

### Step 2 — Create reusable calculations

A **function** lets you use the same calculation for different companies and years. Notice the safeguards: a previous-year revenue or current-year revenue of zero (or less) is rejected, rather than producing a misleading percentage.

```python
def revenue_growth_percent(current_year, previous_year):
    """Revenue growth as a percentage. Assumes comparable positive revenue."""
    if previous_year <= 0:
        raise ValueError("Previous-year revenue must be positive.")
    return (current_year / previous_year - 1) * 100


def operating_margin_percent(operating_profit, revenue):
    """Operating profit / revenue, expressed as a percentage."""
    if revenue <= 0:
        raise ValueError("Revenue must be positive.")
    return operating_profit / revenue * 100
```

### Step 3 — Calculate, compare and interpret

**Revenue growth** measures how much sales changed. **Operating margin** indicates how much operating profit is earned per £100 of sales. A margin change is measured in **percentage points**, not simply percent.

```python
growth = revenue_growth_percent(revenue_2024, revenue_2023)
margin_2023 = operating_margin_percent(operating_profit_2023, revenue_2023)
margin_2024 = operating_margin_percent(operating_profit_2024, revenue_2024)
margin_change = margin_2024 - margin_2023

print(f"Revenue growth, 2023–2024: {growth:.2f}%")
print(f"Operating margin 2023: {margin_2023:.2f}%")
print(f"Operating margin 2024: {margin_2024:.2f}%")
print(f"Margin improvement: {margin_change:.2f} percentage points")
```

### Step 4 — Test that the numbers are correct

An **assertion** is an automated check. These checks are small, but later we will test more complex document extraction and AI outputs too.

```python
assert round(growth, 2) == 13.64
assert round(margin_2023, 2) == 12.73
assert round(margin_2024, 2) == 14.00
assert round(margin_change, 2) == 1.27
print("All four checks passed ✅")
```
